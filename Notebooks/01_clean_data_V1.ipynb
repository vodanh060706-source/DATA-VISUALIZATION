```python
# Chuẩn hóa tên hàng 
mapping = {
    # Thủy sản
    "thủy sản": "Thủy sản",
    "hàng hải sản": "Thủy sản",

    # Rau quả
    "rau quả": "Rau quả",
    "hàng rau quả": "Rau quả",

    # Các mặt hàng giữ nguyên
    "hạt điều": "Hạt điều",
    "cà phê": "Cà phê",
    "chè": "Chè",
    "hạt tiêu": "Hạt tiêu",
    "gạo": "Gạo",
    "cao su": "Cao su",

    # Sắn
    "sắn và sản phẩm của sắn": "Sắn và sản phẩm của sắn",
    "sắn và các sản phẩm từ sắn": "Sắn và sản phẩm của sắn",

    # Xơ
    "xơ, sợi dệt": "Xơ, sợi dệt các loại",
    "xơ, sợi dệt các loại": "Xơ, sợi dệt các loại",
}
# Hàm chuẩn hóahóa
def normalize_item(item):
    item = str(item).strip().lower()
    return mapping.get(item, item.title())
```

## 4. Chuẩn hóa tên cột

Các báo cáo của từng năm có sự khác nhau về tên cột như:

- Tháng 01 và Tháng 1
- Sơ bộ tháng 12
- Sơ bộ năm 2024
- 12 tháng
- Trị giá (1000 USD)

Các tên cột được chuẩn hóa nhằm đảm bảo tất cả các năm có cùng cấu trúc dữ liệu.

```python
# Chuẩn hóa tên cột
def standardize_columns(columns):
    new_columns = []
    for col in columns:
        # NaN
        if pd.isna(col):
            new_columns.append("")
            continue
        col = str(col)
        # Xóa khoảng trắng và xuống dòng
        col = col.replace("\n", " ")
        col = " ".join(col.split())
        # Xóa đơn vị
        col = re.sub(r"\(.*?\)", "", col).strip()
        # STT
        col = col.replace("Stt", "STT")
        # Chuẩn hóa tháng
        col = re.sub(r"Tháng\s*0([1-9])", r"Tháng \1", col)
        # Bỏ chữ "Sơ bộ"
        col = col.replace("Sơ bộ tháng", "Tháng")
        col = re.sub(r"Sơ bộ năm\s+\d{4}", "Cả năm", col)
        # Chuẩn hóa "12 tháng", "cả năm"
        col = col.replace("12 tháng", "Cả năm")
        col = col.replace("cả năm", "Cả năm")
        # Bỏ chữ "Sơ bộ" còn sót
        col = col.replace("Sơ bộ", "").strip()
        # Chuẩn hóa trị giá
        col = col.replace("Trị giá 1000 USD", "Trị giá")
        col = col.replace("Trị giá USD", "Trị giá")
        new_columns.append(col)
    return new_columns
```

## 5. Xây dựng hàm làm sạch dữ liệu

Hàm `clean_file()` thực hiện các bước:

1. Đọc file Excel.
2. Tự động xác định vị trí header.
3. Ghép hai dòng tiêu đề.
4. Chuẩn hóa tên cột.
5. Loại bỏ các cột không cần thiết.
6. Chuẩn hóa tên mặt hàng.
7. Lọc các mặt hàng thuộc phạm vi nghiên cứu.
8. Thêm thông tin năm.

```python
# Xử lí file
keywords = [
    "Thủy sản",
    "Rau quả",
    "Hạt điều",
    "Cà phê",
    "Chè",
    "Hạt tiêu",
    "Gạo",
    "Sắn và sản phẩm của sắn",
    "Cao su",
    "Xơ, sợi dệt các loại"
]
def clean_file(filePath):
    # Đọc file
    df = pd.read_excel(filePath, header=None)
    # Tìm dòng header
    header_row = None
    for i in range(10):
        row = df.iloc[i].astype(str)
        if row.str.contains("Tên hàng|Stt", case=False, na=False).any():
            header_row = i
            break
    if header_row is None:
        raise ValueError(f"Không tìm thấy header trong file: {filePath}")
    # Ghép header
    header1 = df.iloc[header_row].ffill()
    header2 = df.iloc[header_row + 1]
    columns = []
    for h1, h2 in zip(header1, header2):
        h1 = "" if pd.isna(h1) else str(h1).strip()
        h2 = "" if pd.isna(h2) else str(h2).strip()
        if h2 == "":
            columns.append(h1)
        else:
            columns.append(f"{h1}_{h2}")
    # Chuẩn hóa tên cột
    columns = standardize_columns(columns)
    # File 2017 cột thứ 2 đang để trống
    if len(columns) > 1 and columns[1] == "":
        columns[1] = "Tên hàng"
    # Lấy phần dữ liệu
    df = df.iloc[header_row + 3:].reset_index(drop=True)
    # Gán tên cột
    df.columns = columns
    # Xóa các cột không cần
    drop_cols = [
    "STT",
    "",
    "ĐVT",
    "Cả năm_Lượng",
    "Cả năm_Trị giá",
    "12 Tháng_Lượng",
    "12 Tháng_Trị giá"  
    ]
    df = df.drop(
        columns=[c for c in drop_cols if c in df.columns],
        errors="ignore"
    )
    # Chuẩn hóa tên mặt hàng
    item_col = "Tên hàng"
    df[item_col] = (
        df[item_col]
        .astype(str)
        .str.strip() 
        .apply(normalize_item)
    )
    # Chỉ giữ mặt hàng nông sản
    df = df[df[item_col].isin(keywords)].reset_index(drop=True)
    return df
```

## 6. Đọc dữ liệu các năm

Dữ liệu từ năm 2017 đến năm 2025 được đọc tự động bằng vòng lặp và xử lý thông qua hàm `clean_file()`.

```python
# Lưu lại các đường dẫn 
df_full = [] 
for year in range(2017, 2026):
    filePaths = f"/run/media/a356/DATA/DEV/PYTHON/DATA VISUALIZATION/DATA_raw/V1-{year}.xls"
    df = clean_file(filePaths)
    df["Year"] = year
    print(year, df.shape)
    df_full.append(df)
#Ghép tất cả các file lại thành một frameframe
df_full = pd.concat(df_full, ignore_index=True)
# Chuyển đơn vị tinh năm 2017 từ triệu USD về nghìn USD và từ nghìn tấn về tấn 
quantity_cols = [
    col for col in df_full.columns
    if "Lượng" in col
]
df_full.loc[df_full["Year"] == 2017, quantity_cols] *= 1000

value_cols = [
    col for col in df_full.columns
    if "Trị giá" in col
]
df_full.loc[df_full["Year"] == 2017, value_cols] *= 1000
df_full.head(10)
```
