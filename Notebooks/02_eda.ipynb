## 5. Phân tích theo năm

Phân tích sự thay đổi của lượng và trị giá xuất khẩu trong giai đoạn 2017–2025.

Mục tiêu:
- So sánh tổng lượng xuất khẩu giữa các năm.
- So sánh tổng trị giá xuất khẩu giữa các năm.
- Xác định năm có giá trị xuất khẩu cao nhất và thấp nhất.
- Quan sát xu hướng biến động qua thời gian.

```python
# Tổng lượng xuất khẩu theo năm
quantity_cols = [
    col for col in df.columns
    if "Lượng" in col
]

quantity_by_year = (
    df.groupby("Year")[quantity_cols]
      .sum()
      .sum(axis=1)
)
print(quantity_by_year.map(lambda x: f"{x:,.0f} nghìn tấn"))
```

```python
# Tổng trị giá xuất khẩu theo năm
value_cols = [
    col for col in df.columns
    if "Trị giá" in col
]
value_by_year = (
    df.groupby("Year")[value_cols]
      .sum()
      .sum(axis=1)
)
print(value_by_year.map(lambda x: f"{x:,.0f} nghìn USD"))
```

```python
# Năm xuất khẩu cao và thấp nhất
print("Năm có tổng sản lượng cao nhất:", quantity_by_year.idxmax())
print("Năm có tổng sản lượng thấp nhất:", quantity_by_year.idxmin())

print("Năm có tổng trị giá cao nhất:", value_by_year.idxmax())
print("Năm có tổng trị giá thấp nhất:", value_by_year.idxmin())
```

```python
year_summary = pd.DataFrame({
    "Tổng sản lượng": quantity_by_year,
    "Tổng trị giá": value_by_year
})
year_summary_display = year_summary.copy()
year_summary_display["Tổng sản lượng"] = (
    year_summary_display["Tổng sản lượng"]
    .map(lambda x: f"{x:,.0f} nghìn tấn")
)
year_summary_display["Tổng trị giá"] = (
    year_summary_display["Tổng trị giá"]
    .map(lambda x: f"{x:,.0f} nghìn USD")
)
print(year_summary_display)
```

## 6. Phân tích theo mặt hàng

Phân tích lượng và trị giá xuất khẩu của 10 nhóm mặt hàng nông sản
trong giai đoạn 2017–2025.

Mục tiêu:
- So sánh quy mô xuất khẩu giữa các mặt hàng.
- Xác định những mặt hàng có tổng lượng và tổng trị giá xuất khẩu cao nhất.
- Xác định những mặt hàng có đóng góp nổi bật trong tổng kim ngạch xuất khẩu.

```python
# Tổng lượng theo mặt hàng
quantity_by_item = (
    df.groupby("Tên hàng")[quantity_cols]
      .sum()
      .sum(axis=1)
      .sort_values(ascending=False)
)
print(quantity_by_item.map(lambda x: f"{x:,.0f} nghìn tấn"))
```

```python
# Tổng lượng theo mặt hàng
value_by_item = (
    df.groupby("Tên hàng")[value_cols]
      .sum()
      .sum(axis=1)
      .sort_values(ascending=False)
)
print(value_by_item.map(lambda x: f"{x:,.0f} nghìn USD "))
```

```python
# Bảng tóm tắt tổng sản lượng và trị giá 
item_summary = pd.DataFrame({
    "Tổng sản lượng": quantity_by_item,
    "Tổng trị giá": value_by_item
})
item_summary_display = item_summary.copy()
item_summary_display["Tổng sản lượng"] = (
    item_summary_display["Tổng sản lượng"]
    .map(lambda x: f"{x:,.0f} nghìn tấn")
)
item_summary_display["Tổng trị giá"] = (
    item_summary_display["Tổng trị giá"]
    .map(lambda x: f"{x:,.0f} nghìn USD")
)
print(item_summary_display)
```

```python
# Tỷ trọng trị giá 
item_summary_display["Tỷ trọng trị giá (%)"] = (
    item_summary["Tổng trị giá"]
    / item_summary["Tổng trị giá"].sum()
    * 100
).round(2)
display(item_summary_display)
```

```python
# Mặt hàng thấp nhất, cao nhất 
print(
    "Sản lượng cao nhất:",
    quantity_by_item.idxmax()
)
print(
    "Trị giá cao nhất:",
    value_by_item.idxmax()
)
# Sản lượng thấp nhất
print("Sản lượng thấp nhất:")

lowest_quantity = quantity_by_item.sort_values().head(2)

for item, value in lowest_quantity.items():
    print(f"{item}")

print(
    "Trị giá thấp nhất:",
    value_by_item.idxmin()
)
```

## 7. Phân tích theo tháng

Phân tích sự biến động của lượng và trị giá xuất khẩu theo từng tháng
trong giai đoạn 2017–2025.

Mục tiêu:
- Xác định tháng có lượng xuất khẩu cao nhất và thấp nhất.
- Xác định tháng có trị giá xuất khẩu cao nhất và thấp nhất.
- Quan sát xu hướng biến động theo tháng.
- Xem xét khả năng xuất hiện tính mùa vụ trong hoạt động xuất khẩu.

```python
# Tổng sản lượng theo tháng 
monthly_quantity = {}

for month in range(1, 13):
    col = f"Tháng {month}_Lượng"
    
    if col in df.columns:
        monthly_quantity[month] = df[col].sum()
monthly_quantity = pd.Series(monthly_quantity)
print(
    monthly_quantity.map(
        lambda x: f"{x:,.0f} tấn"
    )
)
```

```python
# Tổng trị giá theo tháng 
monthly_value = {}

for month in range(1, 13):
    col = f"Tháng {month}_Trị giá"
    
    if col in df.columns:
        monthly_value[month] = df[col].sum()
monthly_value = pd.Series(monthly_value)
print(
    monthly_value.map(
        lambda x: f"{x:,.0f} nghìn USD"
    )
)
```

```python
# Tháng cao nhất và thấp nhất ở sản lượng và giá trị 
print(
    "Tháng có tổng sản lượng cao nhất: Tháng",
    monthly_quantity.idxmax()
)

print(
    "Tháng có tổng sản lượng thấp nhất: Tháng",
    monthly_quantity.idxmin()
)

print(
    "Tháng có tổng trị giá cao nhất: Tháng",
    monthly_value.idxmax()
)

print(
    "Tháng có tổng trị giá thấp nhất: Tháng",
    monthly_value.idxmin()
)
```

```python
# Tạo bảng văns tắt 
monthly_summary = pd.DataFrame({
    "Tổng sản lượng (tấn)": monthly_quantity,
    "Tổng trị giá (nghìn USD)": monthly_value
})
display(
    monthly_summary.style
    .format("{:,.0f}")
)
```

```python
# Tỷ trọng từng tháng 
monthly_summary["Tỷ trọng trị giá (%)"] = (
    monthly_summary["Tổng trị giá (nghìn USD)"]
    / monthly_summary["Tổng trị giá (nghìn USD)"].sum()
    * 100
).round(2)
display(
    monthly_summary.style.format({
        "Tổng sản lượng (tấn)": "{:,.0f}",
        "Tổng trị giá (nghìn USD)": "{:,.0f}",
        "Tỷ trọng trị giá (%)": "{:.2f}"
    })
)
```

## 8. Phân tích mối quan hệ giữa sản lượng và trị giá

Phân tích mối quan hệ giữa tổng sản lượng và tổng trị giá xuất khẩu
của các mặt hàng trong giai đoạn 2017–2025.

Mục tiêu:
- Xem xét mức độ tương quan giữa sản lượng và trị giá.
- Xác định liệu các mặt hàng có sản lượng cao có đồng thời có trị giá cao hay không.
- Làm cơ sở cho việc trực quan hóa mối quan hệ giữa hai biến.

```python
# Tổng sản lượng và trị giá
df["Tổng Sản lượng"] = df[quantity_cols].sum(axis=1) 
df["Tổng Trị giá"] = df[value_cols].sum(axis=1)
```

```python
# Tính hệ số tương quan giửa sản lượng và trị giá
corralation = df[["Tổng Sản lượng", "Tổng Trị giá"]].corr()
display(corralation)
```

```python
# Tương quan giữa các mặt hàng 
item_corr = {}
for item, group in df.groupby("Tên hàng"):
    item_corr[item] = group["Tổng Sản lượng"].corr(group["Tổng Trị giá"])
item_corr = pd.Series(item_corr).sort_values(ascending=False)
item_corr = (
    item_corr
    .rename("Hệ số tương quan")
    .reset_index()
    .rename(columns={"index": "Tên hàng"})
)
display(
    item_corr
    .style.format({"Hệ số tương quan:": "{:.2f}" }, na_rep="NaN")
)
```

## 9. Phát hiện dữ liệu bất thường

Kiểm tra các giá trị bất thường trong dữ liệu sản lượng và trị giá
xuất khẩu nhằm phát hiện những quan sát có giá trị quá lớn hoặc quá nhỏ
so với phần lớn dữ liệu.

Mục tiêu:
- Phát hiện các giá trị có khả năng là bất thường.
- Xác định các mặt hàng hoặc năm có giá trị khác biệt đáng kể.
- Kiểm tra nguyên nhân trước khi quyết định có xử lý hay không.

```python
# Thống kê dữ liệu 
numeric_col = ["Tổng Sản lượng", "Tổng Trị giá"]
display(df[numeric_col].describe())
```

```python
# Sử dụng phương pháp IRQ để xác định outlier 
Q1 = df[numeric_col].quantile(0.25)
Q3 = df[numeric_col].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR
```

```python
# Tìm dòng outlier
outlier_mask = (
    (df[numeric_col] < lower_bound) |
    (df[numeric_col] > upper_bound)
).any(axis=1)

outliers = df.loc[
    outlier_mask,
    ["Tên hàng", "Year"] + numeric_col
]

display(outliers)
```

```python
print("Số dòng bất thường:", outliers.shape[0])
```

```python
# Nhóm bất thường 
outlier_by_item = (
    outliers["Tên hàng"]
    .value_counts()
)

display(outlier_by_item)
```

### Nhận xét

Các giá trị bất thường được xác định dựa trên phương pháp IQR.
Các quan sát bất thường chưa được loại bỏ mà được giữ lại để kiểm tra
nguyên nhân và đối chiếu với dữ liệu nguồn trước khi thực hiện các bước
xử lý tiếp theo.
